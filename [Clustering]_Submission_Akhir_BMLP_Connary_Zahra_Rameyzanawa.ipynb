```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 1. Memuat Dataset
df = pd.read_csv("data_clustering.csv")

# 2. Menampilkan 5 baris pertama
print("=== HEAD DATASET ===")
display(df.head())

# 3. Menampilkan Informasi Dataset
print("\n=== INFORMASI DATASET ===")
df.info()

# 4. Menampilkan Statistik Deskriptif
print("\n=== STATISTIK DESKRIPTIF ===")
display(df.describe())

# 5. Matriks Korelasi (Fitur Numerik)
plt.figure(figsize=(10, 8))
numeric_cols = df.select_dtypes(include=[np.number]).columns
sns.heatmap(df[numeric_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Matriks Korelasi Fitur Numerik")
plt.tight_layout()  # Mencegah overlap label
plt.show()

# 6. Histogram Seluruh Kolom (Numerik & Kategorikal)
for col in df.columns:
    plt.figure(figsize=(8, 4))
    if df[col].dtype == "object":
        sns.countplot(data=df, x=col, palette="viridis")
        plt.xticks(rotation=45, ha="right")  # Rotasi agar label tidak tumpang tindih
    else:
        sns.histplot(df[col], kde=True, bins=30, color="skyblue")

    plt.title(f"Distribusi Fitur: {col}")
    plt.xlabel(col)
    plt.ylabel("Frekuensi")
    plt.tight_layout()  # Memastikan visualisasi rapi dan label tidak overlap
    plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler

# 1. Cek Missing Values dan Data Duplikat
print("Jumlah Missing Values:\n", df.isnull().sum())
print("\nJumlah Data Duplikat:", df.duplicated().sum())

# 2. Hapus Missing Values dan Data Duplikat
df_clean = df.dropna().drop_duplicates().copy()

# 3. Hapus Kolom ID, Address, dan Date
# (Otomatis menghapus kolom seperti TransactionID, AccountID, DeviceID, IPAddress, MerchantID, TransactionDate)
keywords = [
    "ID",
    "Address",
    "Date",
    "TransactionID",
    "AccountID",
    "DeviceID",
    "IPAddress",
    "MerchantID",
    "TransactionDate",
]
drop_cols = [
    col
    for col in df_clean.columns
    if any(keyword.lower() in col.lower() for keyword in keywords)
]
df_clean = df_clean.drop(columns=drop_cols)

# 4. Penanganan Outlier menggunakan Metode Drop (Syarat Skilled)
num_features = df_clean.select_dtypes(include=["int64", "float64"]).columns
Q1 = df_clean[num_features].quantile(0.25)
Q3 = df_clean[num_features].quantile(0.75)
IQR = Q3 - Q1
outlier_condition = ~(
    (df_clean[num_features] < (Q1 - 1.5 * IQR))
    | (df_clean[num_features] > (Q3 + 1.5 * IQR))
).any(axis=1)
df_clean = df_clean[outlier_condition]

# 5. Binning Data pada Fitur Numerik (Syarat Advanced)
num_col_to_bin = num_features[0]
df_clean[f"{num_col_to_bin}_Binned"] = pd.qcut(
    df_clean[num_col_to_bin], q=3, labels=["Low", "Medium", "High"]
)

# 6. Feature Encoding menggunakan LabelEncoder
label_encoders = {}
cat_features = df_clean.select_dtypes(include=["object", "category"]).columns

for col in cat_features:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))
    label_encoders[col] = le

# 7. Feature Scaling menggunakan StandardScaler (Syarat Skilled)
scaler = StandardScaler()
df_scaled = df_clean.copy()
df_scaled[num_features] = scaler.fit_transform(df_clean[num_features])

print("\n=== Preprocessing Data Selesai ===")
display(df_scaled.head())
```

```python
import joblib
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from yellowbrick.cluster import KElbowVisualizer

# 1. Visualisasi Elbow Method menggunakan KElbowVisualizer
model_kmeans = KMeans(random_state=42, n_init=10)
visualizer = KElbowVisualizer(model_kmeans, k=(2, 10))
visualizer.fit(df_scaled)
visualizer.show()

# Menentukan jumlah cluster optimal secara otomatis
optimal_k = visualizer.elbow_value_ if visualizer.elbow_value_ else 3
print(f"Jumlah Cluster Optimal (K): {optimal_k}")

# 2. Pelatihan Model K-Means
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(df_scaled)

# Menyimpan Model Clustering Utama (Wajib untuk rubrik Basic)
joblib.dump(kmeans, "model_clustering")

# 3. Menghitung Nilai Silhouette Score (Syarat Skilled)
score = silhouette_score(df_scaled, cluster_labels)
print(f"Silhouette Score (K={optimal_k}): {score:.4f}")

# 4. Reduksi Dimensi dengan PCA & Visualisasi Cluster (Syarat Advanced)
pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_scaled)

# Menyimpan Model PCA (Wajib untuk rubrik Advanced)
joblib.dump(pca, "PCA_model_clustering.h5")

# Visualisasi Plot Hasil Clustering 2D
plt.figure(figsize=(8, 6))
plt.scatter(
    df_pca[:, 0], df_pca[:, 1], c=cluster_labels, cmap="tab10", alpha=0.7
)
plt.title("Visualisasi Hasil Clustering (PCA 2D)")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.colorbar(label="Cluster Target")
plt.show()
```

```python
# 1. Menambahkan kolom 'Target' (hasil cluster) ke dataframe
df_scaled["Target"] = cluster_labels
df_clean["Target"] = cluster_labels

# 2. Mengembalikan data ke bentuk awal / sebelum di-scale dan encode (Syarat Skilled & Advanced)
df_inverse = df_clean.copy()
df_inverse[num_features] = scaler.inverse_transform(df_clean[num_features])

for col in cat_features:
    df_inverse[col] = label_encoders[col].inverse_transform(df_clean[col])

# 3. Analisis deskriptif fitur numerik (mean, min, max) per cluster
print("=== ANALISIS DESKRIPTIF FITUR NUMERIK PER CLUSTER ===")
display(df_inverse.groupby("Target")[num_features].agg(["mean", "min", "max"]))

# 4. Analisis deskriptif fitur kategorikal per cluster
print("\n=== ANALISIS DESKRIPTIF FITUR KATEGORIKAL PER CLUSTER ===")
for col in cat_features:
    print(f"\nDistribusi {col}:")
    display(df_inverse.groupby(["Target", col]).size().unstack(fill_value=0))

# 5. Menyimpan data inverse beserta kolom Target ke file CSV (Syarat Advanced)
df_inverse.to_csv("data_clustering_inverse.csv", index=False)
print(
    "\n[SELESAI] Data berhasil disimpan sebagai 'data_clustering_inverse.csv'."
)
```
